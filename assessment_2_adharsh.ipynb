```python
import pandas as pd
df=pd.read_csv("/content/drive/MyDrive/assessment2/student_performance.csv")
print()
```

```python
#A. Data Cleaning
#1. Identify missing values.
print(df.isnull())
#2. Handle missing numerical values appropriately.
numeric_columns = ["Age", "Attendance", "Python_Score", "SQL_Score", "Pandas_Score", "NumPy_Score", "Final_Score"]
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].mean())
```

```python
#B Feature Engineering
score_columns = ["Python_Score","SQL_Score","Pandas_Score","NumPy_Score"]
df["Average_Score"] = df[score_columns].mean(axis=1)
df["Pass_Status"] = np.where(
    df["Average_Score"] >= 50,
    "Pass",
    "Fail")
```

```python
#C. Analysis
#1) Top 5 students based on Average_Score .
print(
    df.nlargest(5, "Average_Score")[
        ["Student_ID", "Name", "Average_Score"]
    ]
)
```

```python
#2)Department having the highest average score.
print(df.groupby("Department")["Average_Score"].mean().sort_values(ascending=False))
```

```python
#3. Average attendance by department.
print(df.groupby("Department")["Attendance"].mean())
```

```python
#4. Number of students who passed and failed.
print(df["Pass_Status"].value_counts())
```

```python
#D. Summary
summary = df.groupby("Department").agg(
    Total_Students=("Student_ID", "count"),
    Average_Score=("Average_Score", "mean"),
    Average_Attendance=("Attendance", "mean"),
    Pass_Count=("Pass_Status", lambda x: (x == "Pass").sum()),
    Fail_Count=("Pass_Status", lambda x: (x == "Fail").sum())
).reset_index()

print(summary)
```

Q17. Sales Data Analysis

```python
import pandas as pd
df=pd.read_csv("/content/drive/MyDrive/assessment2/sales.csv")
print()
```

```python
#A. Data Preparation
sales = pd.read_csv("/content/drive/MyDrive/assessment2/sales.csv")

sales["Order_Date"] = pd.to_datetime(
    sales["Order_Date"],
    dayfirst=True,
    format="mixed"
)

numeric_columns = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

for col in numeric_columns:
    sales[col] = pd.to_numeric(
        sales[col],
        errors="coerce"
    )
    sales[col] = sales[col].fillna(
        sales[col].mean()
    )

print(sales.isnull().sum())
```

```python
#B Sales Analysis
print("Total Sales:", sales["Sales"].sum())
print("Total Profit:", sales["Profit"].sum())
print("Average Sales:", sales["Sales"].mean())
print(
    sales.groupby("Product")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)
```

```python
# C Category Analysis
category_summary = sales.groupby("Category").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Average_Discount=("Discount", "mean")
).sort_values(
    "Total_Sales",
    ascending=False
)

print(category_summary)
```
